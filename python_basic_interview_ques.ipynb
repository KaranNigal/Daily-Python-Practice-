```python
# 7. wp to check if a number is prime or not 

n= int(input('enter the number to check if its prime or not '))

if n>1:
    for i in range(1, int(n**1/2)):
        if(n%i==0):
            print('the number is not prime')
            break
        else:
            print('the number is prime')
else:
    print('enter valid number')
```

```python
# 10.  Write a Python program to calculate the area of a triangle.

s1= int(input('enter the first side'))
s2= int(input('enter the second side'))
s3= int(input('enter the third side'))

s= (s1+s2+s3)/2

area= (s*(s-s1)*(s-s2)*(s-s3))**0.5

print('area of the triangle is: ',area)
```

```python
# 11.  Write a Python program to check Armstrong’s number.

n= int(input('enter the number to check if its armstorng'))
arm=n

size= len(str(n))

sum=0

while n!=0:
    rem = n%10
    sum= sum + (rem**size)
    n= n//10

if sum==arm:
    print('the number is armstrong')

else:
    print('the number is not armstrong')
```

```python
# 12. Write a Python program to check leap year. 

year= int(input('enter the year to check if the year is leap year'))

if (year%400==0) or (year%100!=0) and (year%4==0):
    print('the year is a leap year')
else:
    print('the year is not a leap year')
```
